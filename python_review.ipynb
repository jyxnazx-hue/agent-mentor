#### Shallow copy - copies the outer list, but inner objects are still shared

#### Deep copy - recursively copies everything

use deep copy, to  copy a list of numpy arrays or tensors, because they still share the same memory

````python
import copy

original = [[1, 2], [3, 4]]

shallow = original.copy()
shallow[0][0] = 99
print(original) # [[99, 2], [3, 4]]  ← inner list shared

deep = copy.deepcopy(original)
deep[0][0] = 77
print(original) # [[99, 2], [3, 4]]  ← unchanged
````

#### Tuple unpacking

````python
x, y = (3, 5)
print(x, y) # 3 5

# swap without temp variable
a, b = 1, 2
a, b = b, a
print(a, b) # 2 1

# start unpacking
first, *rest = (1, 2, 3, 4)
print(first) #1
print(rest) # [2, 3, 4]

*init, last = (1, 2, 3, 4)
print(init) # [1, 2, 3]
print(last) # 4
````

#### Dict - it is a mutable, ordered (3.7+), key-value mapping. Keys must be hashable. Values can be anything.

````python
print(metrics["loss"])  # 0.25
print(metrics.get("lr", 0.0)) # 0.0 (default if key missing)
# print(metrics["lr"]) # key error
````

````python
metrics["lr"] = 0.001 # add or update
metrics["loss"] = 0.22 # update
del metrics["epoch"] # delete
````

#### Looping

````python
for key in metrics:
  print(key)

for value in metrics.values():
  print(value)

for key, value in metrics.items():
  print(key, value)
````

#### Dict Comprehensions

```
{key_expr: value_expr for item in iterable if condition}
```

````python
squares = {x: x*x for x in range(5)}
# {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# With condition
even_squares = {x: x*x for x in range(10) if x % 2 == 0}
# {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}

# From zip
names = ["cat", "dog", "bird"]
labels = [0, 1, 2]
class_to_idx = {n: l for n, l in zip(names, labels)}
# {'cat': 0, 'dog': 1, 'bird': 2}

# Invert
idx_to_class = {v: k for k, v in class_to_idx.items()}
# {0: 'cat', 1: 'dog', 2: 'bird'}

# Filter dict
metrics = {"loss": 0.25, "acc": 0.91, "lr": 0.001, "debug": True}
numeric = {k: v for k, v in metrics.items() if isinstance(v, (int, float))}
# {'loss': 0.25, 'acc': 0.91, 'lr': 0.001}

# Transform values
pct = {k: v*100 for k, v in metrics.items() if k in {"loss", "acc"}}
# {'loss': 25.0, 'acc': 91.0}
````
